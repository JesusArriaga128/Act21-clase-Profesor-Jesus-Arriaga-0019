```python
class Profesor:
    def __init__(self, nombre, materia, antiguedad):
        # 3 Atributos
        self.nombre = nombre
        self.materia = materia
        self.antiguedad = antiguedad

    # 3 Funciones (Acciones)
    def explicar_clase(self):
        return f"📢 {self.nombre} está explicando un tema de {self.materia}."

    def calificar_examen(self):
        return f"✍️ El profesor está calificando exámenes de {self.materia}."

    def pasar_lista(self):
        return f"✅ {self.nombre} está pasando lista. ¡Digan presente!"

# --- Creación de Objetos (Instancias) ---

# Profesor 1 👨‍🏫
profesor1 = Profesor("Eliseo Nava", "Matemáticas", "12 años")

# Profesor 2 👩‍🏫

profesor2 = Profesor("Beatriz Ramos", "Historia", "4 años")

# --- Ejecución y Mensajes de Ejemplo ---

print(f"--- 🏫 FICHA DEL DOCENTE 1 ---")
print(f"Nombre: {profesor1.nombre} | Experiencia: {profesor1.antiguedad}")
print(profesor1.explicar_clase())
print(profesor1.pasar_lista())

print("\n" + "--- 🏫 FICHA DEL DOCENTE 2 ---")
print(f"Nombre: {profesor2.nombre} | Experiencia: {profesor2.antiguedad}")
print(f"Nombre: {profesor2.nombre} | Materia: {profesor2.materia}")
print(profesor2.calificar_examen())
print(profesor2.explicar_clase())
```
